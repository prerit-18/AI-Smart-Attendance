# 📓 Course 5: Recurrent Sequence Analysis (LSTM)
This notebook demonstrates how we utilize recurrent neural networks to capture cyclical patterns in student attendance timeline. Explains:
- Sequence vectorization: converting dates into binary checks `[1, 1, 0, 1]`.
- Temporal window mapping using a sliding window $W=5$.
- LSTM architecture execution in Keras.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import models, layers
import matplotlib.pyplot as plt
```

## 1. Sliding Window Formulation Example
We transform a student's binary attendance list into structured training slices.

```python
def make_windows(sequence, window_size=5):
    X, y = [], []
    for i in range(len(sequence) - window_size):
        X.append(sequence[i:i+window_size])
        y.append(sequence[i+window_size])
    return np.array(X), np.array(y)

# Example attendance vector (1 = Present, 0 = Absent)
student_history = [1.0, 1.0, 1.0, 0.0, 1.0, 1.0, 0.0, 1.0, 1.0, 1.0]
X_sample, y_sample = make_windows(student_history, window_size=5)

print(f"Timeline sequence: {student_history}")
for idx in range(len(X_sample)):
    print(f"Sample {idx+1}: Input Window = {X_sample[idx]} -> Next Day Label = {y_sample[idx]}")
```

## 2. LSTM recurrent network compilation

```python
model = tf.keras.Sequential([
    layers.Input(shape=(5, 1)), # 5 days window, 1 feature (binary)
    layers.LSTM(16, return_sequences=False, activation='tanh'),
    layers.Dense(8, activation='relu'),
    layers.Dense(1, activation='sigmoid') # Presence probability output
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```
