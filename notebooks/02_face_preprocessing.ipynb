# 📓 Course 4: Face Detection and Image Preprocessing
This notebook details the computer-vision preprocessing pipeline for facial image normalization. Outlines: 
1. Grayscale conversion for Haar Cascades face boundary proposal.
2. Cropping, padding, and resizing to 160x160 pixels.
3. Blurriness estimation using the variance of the Laplacian.
4. Intensity rescaling.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

## 1. Simulated Image Creation & Face Localization Mock

```python
# Create a dummy 400x400 image containing a circle representing a face
img = np.zeros((400, 400, 3), dtype=np.uint8)
cv2.circle(img, (200, 200), 100, (220, 220, 220), -1) # Draw white face shape
cv2.circle(img, (170, 180), 10, (50, 50, 50), -1)    # Draw eye 1
cv2.circle(img, (230, 180), 10, (50, 50, 50), -1)    # Draw eye 2
cv2.ellipse(img, (200, 230), (40, 20), 0, 0, 180, (50, 50, 50), 3) # Draw mouth

# Mock Bounding Box coordinates (x, y, w, h)
bbox = (100, 100, 200, 200)
x, y, w, h = bbox

# Draw bounding box visualization
vis_img = img.copy()
cv2.rectangle(vis_img, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(vis_img)
plt.title("Detected Face Box")
plt.show()
```

## 2. Crop, Resize and Intensity Normalization

```python
# Crop face
crop = img[y:y+h, x:x+w]

# Resize
resized = cv2.resize(crop, (160, 160), interpolation=cv2.INTER_AREA)

# Normalize pixels to [-1, 1]
normalized = (resized.astype(np.float32) / 255.0 - 0.5) * 2.0

print(f"Cropped shape: {crop.shape}")
print(f"Resized shape: {resized.shape}")
print(f"Normalized min/max: {normalized.min()} / {normalized.max()}")
```

## 3. Sharpness Quality Auditing (Laplacian Variance)

```python
def get_sharpness(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var()

# Compare sharp image vs. blurred image
blurred = cv2.GaussianBlur(img, (25, 25), 0)

sharp_score = get_sharpness(img)
blur_score = get_sharpness(blurred)

print(f"Sharp Image focus score: {sharp_score:.2f}")
print(f"Blurred Image focus score: {blur_score:.2f}")
```
