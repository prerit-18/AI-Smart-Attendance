This notebook details the design and hyperparameter training of our deep convolutional neural networks. We explore:

- Model B: Improved CNN (With Dropout regularizations and BatchNormalization stabilization layers)

```python
import os
import sys
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load Pre-recorded Training Curves
We inspect and plot the validation accuracy and loss profiles generated during system evaluation.

```python
# Resolve workspace paths dynamically
base_dir = os.path.dirname(os.path.dirname(os.path.abspath('.'))) if '__file__' not in globals() else os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
history_dir = os.path.join(base_dir, "project final", "results", "training_history")

baseline_history_path = os.path.join(history_dir, "custom_cnn_baseline_history.json")
improved_history_path = os.path.join(history_dir, "custom_cnn_history.json")

if os.path.exists(baseline_history_path) and os.path.exists(improved_history_path):
    with open(baseline_history_path) as f:
        hist_base = json.load(f)
    with open(improved_history_path) as f:
        hist_imp = json.load(f)
        
    epochs = range(1, len(hist_base['accuracy']) + 1)
    
    # Plot Comparisons
    plt.figure(figsize=(14, 6))
    
    # Train vs Val Loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, hist_base['val_loss'], 'r--', label='Baseline Val Loss')
    plt.plot(epochs, hist_imp['val_loss'], 'g-', label='Regularized Val Loss')
    plt.title('Validation Loss (Course 2 Overfitting Analysis)')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Train vs Val Acc
    plt.subplot(1, 2, 2)
    plt.plot(epochs, hist_base['val_accuracy'], 'r--', label='Baseline Val Acc')
    plt.plot(epochs, hist_imp['val_accuracy'], 'g-', label='Regularized Val Acc')
    plt.title('Validation Accuracy Comparison')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    
    plt.show()
else:
    print("Training history files not found. Please run the evaluation scripts first to output JSON curves.")
```
